<h3>About Dataset</h3>
This is a preprocessed bearing fault dataset under variable rotational speeds. If you find this dataset useful, please cite our work:

"Efficient Bearing Fault Diagnosis for Edge Computing Using Grayscale Spectrograms and Hybrid Neural Model Compression," in IEEE Access, vol. 13, pp. 147494-147510, 2025, doi: 10.1109/ACCESS.2025.3600678.

<h3>Bearing specification: </h3>
Pitch diameter: 46.5 mm
Rolling element diameter: 9 mm
Contact angle: 0°
Number of elements: 13.
Data acquisition:
Signal type: acoustic emission
Sampling frequency: 250 kHz
Crack size: 12 mm, 6 mm, 3 mm
Tested RPM: 250, 300, 350, 400, 450, 500.

<h3>Preprocessing technique:</h3>
Short-time Fourier transform (STFT).

<h1>RADAR DATA FOR MACHINE FAULT DETECTION</h1>
SELCUK UNIVERSITY REMOTE SENSING LAB. RADAR DATA FOR FAULT DIAGNOSIS

<h3>About Dataset: </h3>
The dataset includes raw radar signals recorded for 659 separate experiments and power spectral density signals, spectrogram images, and scalogram images generated from these raw signals. The experiments include six distinct fault conditions and healthy condition data for an induction motor operated under different operating conditions. Operating conditions are varied speeds ranging from 500 to 1500 rpm and load levels ranging from 0 to 2.5 Nm. The faults consist of winding faults, rotor bar cracks, bearing faults, horizontal and vertical misalignment, and imbalance faults.

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("yunusemreacar1/radar-data-for-machine-fault-detection", output_dir="./dataset/radar-data-for-machine-fault-detection")

print("Path to dataset files:", path)
```
